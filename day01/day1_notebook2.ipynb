```python
import numpy as np
import matplotlib.pyplot as plt

# Create a blank 400x400 image
image = np.zeros((400, 400), dtype=np.uint8)

# 1. Draw the left and right vertical stems
image[100:300, 100:125] = 255  # Left stem
image[100:300, 275:300] = 255  # Right stem

# 2. Draw the diagonal strokes for the inner V-shape of the M
for i in range(100):
    r = 100 + i
    # Left diagonal (from top-left stem down to center)
    c_left = int(125 + (200 - 125) * (i / 100))
    image[r, c_left - 12:c_left + 12] = 255
    
    # Right diagonal (from center up to top-right stem)
    c_right = int(200 + (275 - 200) * (i / 100))
    image[(299-r), c_right - 12:c_right + 12] = 255

# Display the image
plt.imshow(image, cmap="gray")
plt.axis("off")
plt.show()
```
