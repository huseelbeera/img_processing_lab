```python
import easyocr
import cv2
import numpy as np
from pdf2image import convert_from_path
import matplotlib.pyplot as plt

# ---------------------------------------------------------
# Step 0: Direct Poppler Bin Path Override
# ---------------------------------------------------------
POPPLER_BIN_PATH = r"C:\Poppler\Release-26.09.0-0\poppler-26.09.0\Library\bin"

# ---------------------------------------------------------
# Step 1: Initialize EasyOCR Model
# ---------------------------------------------------------
# 'ar' for Arabic text, 'en' for numbers and mixed English terms
print("Loading EasyOCR model...")
reader = easyocr.Reader(['ar', 'en'], gpu=False) 

# ---------------------------------------------------------
# Step 2: Convert PDF to High-Resolution Images
# ---------------------------------------------------------
pdf_path = r"C:\Users\PC-LAB1\img_processing_lab\day10\العدد-الأول-من-الجريدة-الرسمية.pdf"

print("Converting PDF pages to images (300 DPI)...")
pages = convert_from_path(pdf_path, dpi=300, poppler_path=POPPLER_BIN_PATH)
print(f"Total Pages Converted: {len(pages)}")

# ---------------------------------------------------------
# Step 3: Contrast Enhancement Preprocessing
# ---------------------------------------------------------
def preprocess_arabic_image(pil_img):
    """
    Converts PIL image to OpenCV grayscale and applies CLAHE contrast adjustment.
    Preserves delicate Arabic dots and letter connections.
    """
    img = np.array(pil_img)
    img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)

    # 1. Convert to Grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # 2. Contrast Limited Adaptive Histogram Equalization (CLAHE)
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    enhanced = clahe.apply(gray)

    return enhanced

# ---------------------------------------------------------
# Step 4: Extract Text Page-by-Page
# ---------------------------------------------------------
extracted_results = []

for page_num, page_img in enumerate(pages):
    print(f"\nProcessing Page {page_num + 1} of {len(pages)}...")

    # Preprocess image
    cleaned_img = preprocess_arabic_image(page_img)

    # Perform OCR
    # paragraph=True groups words into cohesive paragraphs/lines
    results = reader.readtext(cleaned_img, paragraph=True)

    # Combine extracted text lines for this page
    page_text = "\n".join([text for bbox, text in results])
    extracted_results.append(f"--- PAGE {page_num + 1} ---\n" + page_text)
    print(f"Done with Page {page_num + 1}!")

# ---------------------------------------------------------
# Step 5: Save Extracted Text to File
# ---------------------------------------------------------
full_text = "\n\n".join(extracted_results)

output_filename = "extracted_arabic_text.txt"
with open(output_filename, "w", encoding="utf-8") as f:
    f.write(full_text)

print(f"\nSuccess! Full document saved to: {output_filename}")
print("=" * 50)
print("Sample Extracted Text (First 500 characters):")
print("=" * 50)
print(full_text[:500])
```
