```python
import cv2
from ultralytics import YOLO

# Load YOLO model
model = YOLO("yolo11n.pt")

# Open video file
cap = cv2.VideoCapture(r"C:\Users\PC-LAB1\img_processing_lab\videos\dhaka_traffic.mp4")

# COCO Class IDs
CLASS_PERSON = 0
CLASS_CAR = 2
CLASS_MOTORCYCLE = 3
CLASS_BUS = 5

frame_number = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    frame_number += 1

    # Perform detection
    results = model(frame)

    # Annotate frame with standard bounding boxes
    annotated = results[0].plot()

    # Extract detected class IDs
    boxes = results[0].boxes
    if boxes is not None and len(boxes) > 0:
        classes = boxes.cls.cpu().numpy().astype(int)
    else:
        classes = []

    # Count specific target classes
    person_count = (classes == CLASS_PERSON).sum()
    car_count = (classes == CLASS_CAR).sum()
    motorcycle_count = (classes == CLASS_MOTORCYCLE).sum()
    bus_count = (classes == CLASS_BUS).sum()

    # Prepare overlay text lines
    text_lines = [
        f"Frame: {frame_number}",
        f"Cars: {car_count}",
        f"Persons: {person_count}",
        f"Buses: {bus_count}",
        f"Motorcycles: {motorcycle_count}",
    ]

    # Display counts on top left corner
    x, y = 20, 35
    line_height = 30

    for i, line in enumerate(text_lines):
        # Draw dark background outline/shadow for readability
        cv2.putText(
            annotated,
            line,
            (x, y + i * line_height),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (0, 0, 0),
            4,
            cv2.LINE_AA,
        )
        # Draw white foreground text
        cv2.putText(
            annotated,
            line,
            (x, y + i * line_height),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (255, 255, 255),
            2,
            cv2.LINE_AA,
        )

    # Show the annotated image
    cv2.imshow("YOLO Detection", annotated)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
